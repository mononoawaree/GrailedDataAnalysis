Join embeddings to a parquet on id

```python
import numpy as np
import polars as pl


def open_clip_v32_embbeds(df: pl.LazyDataFrame):
    embbeds = np.load('../data/embeddings.npy', mmap_mode='r')
    oks = np.load('../data/ok.npy', mmap_mode='r')
    ids = pl.Series(df.select('id').collect())
    lookup = pl.LazyFrame({
        'id': ids,
        'open_clip_v32_embbeds': embbeds,
    })
    df = df.join(lookup, on='id', how='left')
```
